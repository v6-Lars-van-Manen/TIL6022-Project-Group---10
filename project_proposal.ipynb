# Motivation and introduction

The railway between Amersfoort, Apeldoorn and Deventer links the Randstad to the eastern Netherlands, and its services are regularly disrupted. On 15 September 2026 (Prinsjesdag), objects deliberately placed on the tracks at 35 or more locations disrupted train traffic in the centre and east of the country (NOS, 2026). The Amersfoort–Deventer corridor was hit as well: that evening, a new obstruction near Teuge halted trains between Apeldoorn and Deventer once more (NOS, 2026).

When such a line fails, travellers switch to other routes, the bus or the car, or cancel their trip, which moves the pressure elsewhere in the transport network. The A1 motorway runs roughly parallel to this railway, making it the most likely road to absorb travellers who switch from train to car. This project uses observed road traffic data to investigate whether the rail disruptions of 15 September 2026 left a measurable trace in traffic intensity and vehicle speed on the A1.

# Research objective

**Research question**

To what extent did the railway disruptions on the Amersfoort–Apeldoorn–Deventer corridor on 15 September 2026 lead to measurable changes in road traffic intensity and average vehicle speed on the parallel A1 corridor?

**Objectives**

- Construct a normal traffic profile for the A1 between Amersfoort and Deventer from historical NDW measurements.
- Quantify how traffic intensity and average speed on 15 September 2026 deviated from that profile, per location, direction and hour.
- Determine whether these deviations coincide in time and place with the rail disruptions.

Further sabotage of the railway network cannot be ruled out. Understanding how the road network responds when a rail corridor fails helps road authorities such as Rijkswaterstaat prepare targeted measures for the A1 and for other corridors that are vulnerable to rail disruptions.

*NDW Dexter* (n.d.): <https://dexter.ndw.nu/opendata/overview>

The NDW (Nationaal Dataportaal Wegverkeer, or National Data Portal Road Traffic) makes mobility data trustworthy, accessible and useful. It does this together with governmental partners such as Rijkswaterstaat, the Dutch provinces and Dutch municipalities, and publishes many types of data through several applications. For this project, the Dexter application will be used as a data source.

Dexter allows the user to select a timeframe of up to 366 days, the days of the week to include, and a specific measurement location. It then delivers a report containing the number of vehicles counted in every hour of the day and the average speed during every hour of the day. # Data pipeline

The data analysis pipeline will be used to determine whether the railway disruption on 15 September 2026 resulted in measurable changes in road traffic along the A1 corridor between Amersfoort, Apeldoorn and Deventer. Historical traffic measurements from NDW Dexter will be used to compare the disruption day with normal traffic conditions.

### 1. Data collection

Traffic data will be collected from several NDW measurement locations along the A1 between Amersfoort, Apeldoorn and Deventer. Where possible, measurement points in both driving directions and at multiple locations along the corridor will be included.

For each measurement point, the main variables are:

- traffic intensity in vehicles per hour;
- average vehicle speed in km/h;
- measurement location and direction;
- time of day.

The disruption day, Tuesday 15 September 2026, will be analysed separately. To construct a representative baseline, the reference dataset will consist of all Tuesdays during the year preceding the disruption that fall outside Dutch school holiday periods. Tuesdays are selected because traffic patterns vary systematically between weekdays, while excluding school holidays reduces differences caused by unusual travel demand.

### 2. Data cleaning and pre-processing

The NDW datasets will be imported into Python and converted into a consistent format. Measurements from the different locations and dates will be combined into one dataset containing location, direction, date, time, traffic intensity and average speed.

Missing or technically unreliable measurements will be identified and removed where necessary. The reference data will then be filtered so that only eligible Tuesdays outside school holidays remain. Measurements will be aggregated into consistent hourly time intervals, allowing the disruption day and the baseline to be compared directly.

### 3. Baseline construction

For each measurement location, direction and hour of the day, the traffic measurements of all selected reference Tuesdays will be combined to create a normal traffic profile.

The mean traffic intensity and mean vehicle speed will be calculated for each hourly interval. This produces an expected value for a normal Tuesday, for example the average traffic intensity between 08:00 and 09:00 across all eligible Tuesdays in the previous year.

Using a full year of comparable Tuesdays rather than a single reference day reduces the influence of normal day-to-day variation and provides a more stable baseline.

### 4. Comparative analysis

Traffic conditions on 15 September 2026 will be compared with the baseline for the same location, direction and hour. For traffic intensity, both the absolute and the relative difference will be calculated:

$$\Delta I = I_{\text{disruption}} - I_{\text{baseline}}$$

*Formula 1: absolute difference in traffic intensity*

$$\%\Delta I = \frac{I_{\text{disruption}} - I_{\text{baseline}}}{I_{\text{baseline}}} \times 100$$

*Formula 2: relative difference in traffic intensity*

The same calculations will be performed for average vehicle speed.

A higher-than-normal traffic intensity may indicate additional road demand during the rail disruption, while a decrease in average speed may indicate additional congestion. Both variables will therefore be analysed together.

### 5. Spatial and temporal analysis

The calculated differences will be compared across locations and throughout the day. This makes it possible to investigate both when abnormal traffic conditions occurred and where along the corridor the largest deviations were observed.

Particular attention will be given to the time periods in which rail services were disrupted. Comparing several measurement locations also allows the analysis to determine whether changes were limited to specific sections of the A1 or occurred throughout the corridor.

### 6. Visualization and interpretation

The results will be visualized using:

- time-series plots comparing the disruption day with the normal Tuesday baseline;
- plots of the percentage change in traffic intensity and average speed;
- comparisons between different measurement locations and directions;
- where useful, a geographical visualization showing the magnitude of the observed changes along the corridor.

These visualizations will be used to assess whether road traffic on 15 September differed systematically from typical Tuesday traffic and whether these differences coincide with the railway disruption.
